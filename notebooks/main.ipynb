```python
from ultralytics import YOLO

# Load YOLOv8 model
model = YOLO("yolov8m.pt")  # Choose larger models if needed

model.info()  # Display architecture details and training hyperparameters
```

```python
# Train using small dataset
model.train(data="dataset_small/dataset_small.yaml", epochs=20, imgsz=640, batch=8)

# Save trained model path
model_path = "runs/detect/train/weights/best.pt"
print(f"Training completed! Model saved at {model_path}")
```

```python
# Load the trained model
model = YOLO("runs/detect/train/weights/best.pt")

# Run inference on a new invoice image
results = model("dataset_small/images/train/Template1_Instance135.jpg", save=True, conf=0.3)

print("Inference done! Processed image saved.")
```

```python
from ultralytics import YOLO
import json

# Load the trained model
model = YOLO("runs/best.pt")

# Run inference on a new invoice image
results = model("dataset/images/test/Template48_Instance99.jpg")

output_data = []
for result in results:
    for box in result.boxes.data:
        x1, y1, x2, y2, conf, cls = box.tolist()
        label = result.names[int(cls)]  # Get class name

        output_data.append({
            "label": label,
            "confidence": conf,
            "bounding_box": {
                "x1": x1,
                "y1": y1,
                "x2": x2,
                "y2": y2
            }
        })

# Save as JSON
with open("output.json", "w") as f:
    json.dump(output_data, f, indent=4)

print("Bounding box details saved to output.json")
result.show()
```
